```python
def read_file(filename):
    fp = open(filename,'r')
    data = fp.read()
    return data

def file_write(filename,data):
    fp = open(filename,'w')
    for item in data:
        fp.write(item+'\n')
    fp.close()

def mobile_no_validation(data):
    valid_mob = []
    for word in data.split():
        if word.isnumeric():
            if len(word) == 10:
                if word[0] in '789':
                    valid_mob.append(word)
                    
    return valid_mob

def email_validation(data):
    valid_email = []
   
    for word in data.split():
        if word.count('@') == 1:
            if word.endswith('@gmail.com'):
                valid_email.append(word)
            
    return valid_email

def invalid_data(data):
    invalid_file = []
    for word in data.split():
        if not word.isnumeric() and  not word.endswith('@gmail.com'):
            invalid_file.append(word)
            
    return invalid_file
            
            
def main():
    file = 'sample123.txt'
    fdata = read_file(file)
    valid_mob = mobile_no_validation(fdata)
    valid_email = email_validation(fdata)
    invalid_Data = invalid_data(fdata) 
    
    file_write('valid_mobile.txt',valid_mob)
    file_write('valid_email.txt',valid_email)
    file_write('Invalid_data.txt',invalid_Data)
    
    
    
if __name__=='__main__':
    main()
```

```python
def file_read(filename):
    with open(filename) as fp:
        data = fp.read()
        return data

def count(filename):
    numeric = 0
    specialchar = 0

    characters = len(filename)
    words = len(filename.split())
    lines = len(filename.splitlines())

    for ch in filename:
        if ch.isdigit():
            numeric += 1
        elif not ch.isalnum() and ch != " ":
            specialchar += 1
    
    result = {
    "characters": characters,
    "words": words,
    "numeric": numeric,
    "special_characters": specialchar,
    "lines": lines
    }

    return result  
def write_file(result):
    with open("sample_output.txt","w") as fp:
        fp.write(str(result))

def main():
    filename = "Sample123.txt"
    fdata = file_read(filename)
    result = count(fdata)
    write_file(result)
    print(result)

if __name__ == '__main__':
    main()
```

```python
Assignment No.02: #Date - 10/09/2026
input file- 
sample.txt-1

    output : 3 files
    1.mobile data -- mobile.txt
    2.email-- email.txt
    3.invalid data
    4.character data-- character.txt
    5.specialcharacter-- specialcharacter.txt
    6.numeric data-- numeric.txt
```

```python
def read_file(filename):
    with open(filename) as fp:
        data = fp.read()
    return data

def valid_mobile(data):
    valid_mobile = []
    invalid_data = []
    for word in data.split():
        if word.isnumeric():
            if len(word) == 10 and word[0] in ('678'):
                valid_mobile.append(word)
            else:
                invalid_data.append(word)
            
    return valid_mobile,invalid_data
            

def valid_email(data):
    valid_email = []
    invalid_data = []
    for word in data.split():
        if word.count('@') == 1:
            if word.endswith("@gmail.com"):
                valid_email.append(word)
        elif '@' in word:
            invalid_data.append(word)
            
    return valid_email,invalid_data


def character_data(data):
    character_data = []
    for word in data.split():
        if word.isalpha():
            character_data.append(word)

    return character_data

def special_char(data):
    spec = []
    for word in data.split():
        if not word.isalnum():
            spec.append(word)
    return spec

def numeric_data(data):
    numeric_data = []
    for word in data.split():
        if word.isnumeric():
            numeric_data.append(word)
    return numeric_data


def write_file(filename,data):
    with open(filename,"w") as fp:
        data = fp.write(str(data))
    return data

def  main():
    data = read_file("sample123.txt")
    valid_mob,invalid_data_mobile = valid_mobile(data)
    valid_email_data,invalid_data = valid_email(data)
    char_data = character_data(data)
    num_data = numeric_data(data)
    spec_data = special_char(data)
    invalid_data = invalid_data_mobile + invalid_data
    write_file("valid_mobile.txt",valid_mob)
    write_file("Valid_email.txt",valid_email_data)
    write_file("invalid_data.txt",invalid_data)
    write_file("character_data.txt",char_data)
    write_file("special_char_data.txt",spec_data)
    write_file("numeric_data.txt",num_data)
    
    
if __name__ == '__main__':
    main()
```
