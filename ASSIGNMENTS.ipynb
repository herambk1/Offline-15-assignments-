```python
#open sample file
def count_file_details():
    with open(r"C:\Users\Admin\Desktop\PYTHON\\Text Files\\sample.txt","r") as file:
        data = file.read()
    #1.Count Characters
    char = len(data)

    #2.Count Words 
    words = len(data.split())

    #3.Count SpecialCharacters
    specialchar= 0
    for ch in data:
        if not ch.isalnum() and not ch.isspace():
            specialchar+=1

    #4.Count Lines
    lines = len(data.splitlines())

    #5.Count Numbers
    numbers = 0 
    for ch in data:
        if ch.isdigit():
            numbers+=1

    #6.Result
    result = {
        "Char": char,
        "Words": words,
        "SpecialCharacters": specialchar,
        "Lines": lines,
        "Numbers": numbers
    }

    #7.Output in sample.output file
    with open(r"C:\Users\Admin\Desktop\PYTHON\\Text Files\\sample_output.txt","w") as file:
            file.write("Character : " + str(char) +'\n')
            file.write("Words : " + str(words) +'\n')
            file.write("SpecialCharacter : " + str(specialchar) +'\n')
            file.write("Lines : " + str(lines) +'\n')
            file.write("Numbers : " + str(numbers) +'\n')
    return result

# Calling the function
result = count_file_details()

print(result)    
print("Output file created successfully.")
```

```python
Assignment No.02: #Date - 10/09/2026
input file- 
sample.txt-1

    output : 3 files
    1.mobile data -- mobile.txt
    2.email-- email.txt
    3.invalid data
    output : 3 files
    4.character data-- character.txt
    5.specialcharacter-- specialcharacter.txt
    6.numeric data-- numeric.txt
```

```python
def file_read(filename):
    fp = open(filename)
    data = fp.read()
    fp.close()
    return data
    
def mobile_filteration(filedata):
    valid_mobile = []
    invalid_mobile = []
    for words in filedata.split():
        if words.isnumeric():
            if len(words) == 10:
                if words.startswith('7') or words.startswith('8') or words.startswith('9'):
                    valid_mobile.append(words)
                else:
                    invalid_mobile.append(words)

    return valid_mobile, invalid_mobile
    
def email_filterartion(filedata):
    valid_email = []
    invalid_email = []
    for words in filedata.split():
        if "@" in words:
            if words.endswith("@gmail.com"):
                valid_email.append(words)
            else:
                invalid_email.append(words)
    return valid_email, invalid_email

def characters(filedata):
    characters = []
    for words in filedata.split():
        for ch in words:
            if ch.isalpha():
                characters.append(ch)
    return characters

def specialcharacters(filedata):
    specialcharacters = []
    for words in filedata.split():
        for ch in words:
            if not ch.isalnum():
                specialcharacters.append(ch)
    return specialcharacters

def numbers(filedata):
    numbers = []
    for words in filedata.split():
        for ch in words:
            if ch.isdigit():
                numbers.append(ch)
    return numbers

def file_write(filename, data):
    fp = open(filename, "w")
    for item in data:
        fp.write(item + '\n')
    fp.close()

def main():
    filename = r"C:\Users\Admin\Desktop\PYTHON\\Text Files\\sample.txt"
    fdata = file_read(filename)

    valid_mobile, invalid_mobile = mobile_filteration(fdata)
    valid_email, invalid_email = email_filterartion(fdata)

    character_data = characters(fdata)
    specialcharacter_data = specialcharacters(fdata)
    number_data = numbers(fdata)

    file_write(r"C:\Users\Admin\Desktop\PYTHON\\Text Files\\valid_mobile.txt", valid_mobile)
    file_write(r"C:\Users\Admin\Desktop\PYTHON\\Text Files\\valid_email.txt", valid_email)
    file_write(r"C:\Users\Admin\Desktop\PYTHON\\Text Files\\invalid.txt", invalid_mobile + invalid_email)

    file_write(r"C:\Users\Admin\Desktop\PYTHON\\Text Files\\characters.txt", character_data)
    file_write(r"C:\Users\Admin\Desktop\PYTHON\\Text Files\\specialcharacters.txt", specialcharacter_data)
    file_write(r"C:\Users\Admin\Desktop\PYTHON\\Text Files\\numbers.txt", number_data)


if __name__ == "__main__":
    main()
```

```python
import csv

def file_read(filename):
    with open ("{}".format(filename),"r",newline="") as fp:
        r = csv.DictReader(fp)
        data = list(r)
    return data
```

```python
import csv
#from utilis.read_csv_file import file_read

def create_department_dictionary(data):
    dept_dict = {}

    for line in data:
        dept = line['dept']
        if dept not in dept_dict:
            dept_dict[dept] = []
            dept_dict[dept].append(line)
    return dept_dict

def file_write(dept, data):
    filename = ("{}_records.csv".format(dept))
    with open(filename, "w", newline="") as fp:
        fn = data[0].keys()
        w = csv.DictWriter(fp, fieldnames=fn)
        w.writeheader()
        w.writerows(data)

def main():
    filename = r'C:\Users\Admin\PycharmProjects\CSV Operation\empsal_part_buck.csv'
    data = file_read(filename)
    dept_dict = create_department_dictionary(data)
    for dept, records in dept_dict.items():
        file_write(dept, records)

if __name__ == '__main__':
    main()
```

```python
import json
import csv

def read_json_file(filename):
    with open("{}.json".format(filename), "r", newline="") as fp:
        return json.load(fp)
        
def get_columns(file):
    columns = {}
    for record in file:
        for key in record:
            columns[key] = 1
    return list(columns.keys())

def to_csv(keys, file):
    filename = ("{}.csv".format("emp2_output"))
    with open(filename, "w", newline="") as fp:
        writer = csv.DictWriter(fp, fieldnames=keys)
        writer.writeheader()
        writer.writerows(file)
        
if __name__ == '__main__':
    file = read_json_file(r"C:\Users\Admin\Desktop\PYTHON\emp2")
    keys = get_columns(file)
    to_csv(keys, file)
    print(keys)
```

```python
import yaml
import csv

def yaml_to_csv(input_file, output_file):
    with open(input_file, "r") as file:
        data = yaml.safe_load(file)

    records = data["employees"]
    fieldnames = records[0].keys()

    with open(output_file, "w", newline="") as file:
        writer = csv.DictWriter(file, fieldnames=fieldnames)

        writer.writeheader()
        writer.writerows(records)

    print("CSV file created successfully!")
    
def main():
    input_file = r"C:\Users\Admin\Desktop\PYTHON\emp_yaml.yml"
    output_file = r"C:\Users\Admin\Desktop\PYTHON\emp_yaml_csv.csv"
    yaml_to_csv(input_file, output_file)
    
if __name__ == "__main__":
    main()
```

```python
import yaml
import json
import pymysql

# Read YAML file
with open(r"C:\Users\Admin\Desktop\PYTHON\Yaml connect to DB and Cred in Config.json\student.yaml", "r") as fp:
    data = yaml.safe_load(fp)

# Read database credentials from config.json
with open(r"C:\Users\Admin\Desktop\PYTHON\Yaml connect to DB and Cred in Config.json\config.json", "r") as fp:
    config = json.load(fp)

# Connect to MySQL database
conn = pymysql.connect(
    host=config["host"],
    port=config["port"],
    user=config["user"],
    password=config["password"],
    database=config["database"]
)

cur = conn.cursor()

# Insert YAML data into student table
for line in data["students"]:
    query = "insert into student values('{}','{}','{}')".format(line['id'], line['name'], line['city'])
    cur.execute(query)

conn.commit()
conn.close()
```
