```python
def mygen(start,end):
    i = start
    while i<=end:
        yield i
        i+=1
```

```python
obj = mygen(5,500)
```

```python
for i in obj:
    if i ==100:
        break
    else:
        print(i)
```

```python
p = ['56','7','7','76']
```

```python
def convert_into_int_rev(plist):
    pl =[]
    for element in plist:
        pl.append(int(element))
    pl.reverse()
    return pl
```

```python
convert_into_int_rev(p)
```
