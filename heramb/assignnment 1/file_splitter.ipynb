```python
def read_file(filename):
    with open(filename, "r") as file:
        return file.read()

def is_char(data):
    count = 0
    for ch in data:
        if ch.isalpha():
            count+=1
    return count
    

def is_num(data):
    count = 0
    for ch in data:
        if ch.isnumeric():
            count +=1
    return count

def is_special(data):
    count = 0
    for ch in data:
        if not ch.isalnum() and not ch.isspace():
            count+=1
    return count



if __name__ == "__main__":

    data = read_file("sample.txt")

    dict ={
        "numbers":is_num(data),
        "characters":is_char(data),
        "Special Chars": is_special(data)
    }
    print(dict)
  
```
