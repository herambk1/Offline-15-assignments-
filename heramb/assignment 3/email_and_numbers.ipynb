```python
def read_file(filename):
    with open(filename,"r")as file:
        return file.read()

def is_email(data):
    invalid_emails=[]
    with open("emails.txt","w") as file:
        email = data.split()
        for word in email:
            if '@' in word:
                parts = word.split('@')
                if len(parts) ==2:
                    username,domain= parts 
                    if username.isalnum():
                        if not domain.startswith('.') and not domain.endswith('.'):
                            if '.com' in domain:
                                file.write(word+'\n')
                            else:
                                invalid_emails.append(word)
                        else:
                            invalid_emails.append(word)
                    else:
                        invalid_emails.append(word)
                else:
                    invalid_emails.append(word)
            else:
                    invalid_emails.append(word)
                
    return invalid_emails       
      

def is_number(data):
    validph=[]
    invalidph=[]
 
    with open("Phone_no.txt","w") as phfile:
      
        phno = data.split()
        for ch in phno:
            if ch.isnumeric() and len(ch)==10:
                if ch.startswith('7') or ch.startswith('9') or ch.startswith('8'):   #6123456789 7123456789 8123456789
                    phfile.write(ch +'\n')
                else:
                   invalidph.append(ch) 
            elif ch.isnumeric():
                invalidph.append(ch)
      
    return invalidph
    

def invalid_email_and_phoneno(var):
    with open("invalid.txt","w") as file:
        for ch in var:
            file.write(ch + '\n')
            
    

if __name__ == "__main__":
    data = read_file("input.txt")
    email = is_email(data)
    phno = is_number(data)
    phno.extend(email)
    invalid_email_and_phoneno(phno)
    #print(temp,email,phno)
    
    
```
