```python
import csv

def read_file(filename):
    with open(filename,"r") as file:
        data = list(csv.DictReader(file))
    return data


def get_dept(data):
    dept = set()
    for d in data:
        dept.update({d['department']})
    return dept
            
    
def sort_dept_data(dept,data):
    for char in dept:                                                              
        sorted_data = process_data(char,data)                                      
        write_Dept_file(char,sorted_data)
            
def process_data(dept,data):
    records=[]
    for record in data:
        if record['department']==dept:                         
            records.append(record)
    return records
        

def write_Dept_file(deptname, data):
    filename = '{}.csv'.format(deptname)
    with open(filename,"w") as file:
        headers = data[0].keys()
        w = csv.DictWriter(file, fieldnames=headers)
        w.writeheader()
        w.writerows(data)


if __name__ == "__main__":
    data = read_file("emp1.csv")
    dept = get_dept(data)
    sort_dept_data(dept,data)    
    # wriet_Dept_file(deptName,dept)
        # for d in data:
    #     print(d)
    #Operations HR Finance IT Sales
    for d in dept:
        print(d)

    
    
```
