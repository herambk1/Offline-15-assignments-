```python
import json
import csv
```

```python
def readjson_file(filename):
    with open(filename,"r") as file:
        return json.load(file)

def convertToCsv(keys, file):
    with open("users.csv","w", newline="") as fp:
        writer = csv.DictWriter(fp, fieldnames=keys)
        writer.writeheader()
        writer.writerows(file)

def get_columns(file):
    return list(  
        dict.fromkeys(
            key
            for record in file
            for key in record.keys()
        )
    )
    

if __name__ =='__main__':
    file = readjson_file(r"D:\Downloads\users.json")
    keys = get_columns(file)
    convertToCsv(keys,file)
    #convertToCsv(file)
    #print(file)
    #keys = file[0].keys()
    print(keys)
```
