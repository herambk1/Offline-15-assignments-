```python
def even_odd_num(plist):
    even =[]
    odd =[]
    for num in plist:
        if num%2==0:
            even.append(num)
        else:
            odd.append(num)

    return even,odd
```

```python
p=[45,32,65,34,342,45,65,76,32,21,43,54]
e,o=even_odd_num(p)
print(e,o)
```

```python
def get_input_from_users():
    x = int(input("Enter the marks for sub1:"))
    y = int(input("Enter the marks for sub2:"))
    z = int(input("Enter the marks for sub3:"))
    return x,y,z

def calculate_total(x,y,z):
    return (x+y+z)

def calculate_avg(t1):
    return (t1)/3

def calculate_per(total):
    per = (total/100)*3
    return per

def categories_grade(per):
    if per >35 and per<60:
        return "C"
    elif per>60 and per<80:
        return "B"
    elif per>80 and per<100:
        return "A"
    else:
        return "Fail"
        

def main():
    s1,s2,s3 = get_input_from_users()
    total = calculate_total(s1,s2,s3)
    avg = calculate_avg(total)
    per = calculate_per(total)
    grade = categories_grade(per)

    print("Subject Marks:", s1, s2, s3)
    print("Total:",total)
    print("Avg:" ,avg)
    print("Percentage:",per)
    print("Grade:",grade)


    
if __name__ == '__main__':
    main()

    
```

```python
def db_decor(funct):
    def inner():
        print("connection established")
        funct()
        print("connection terminated")
    return inner


@db_decor
def copper1():
    print("Data Transfer for copper")


@db_decor
def silver1():
    print("Data Transfer for silver")


@db_decor
def gold1():
    print("Data Transfer for gold")


```

```python
copper1()
silver1()
gold1()
```
