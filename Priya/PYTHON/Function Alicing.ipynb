```python
def login(funct):
    def inner():
        print("connecting...server")
        print("connected")
        funct()
        print("disconected")
    return inner
```

```python
def db_decorator(funct):
    def inner():
        print("connection establishment")
        funct()
        print("connection terminated")
    return inner 

@db_decorator    
def db_copper():
    print("data transfer for copper")
    
@db_decorator
def db_silver():
    print("data transfer for silver")

@db_decorator
def db_gold():
    print("data transfer for gold")
    
```

```python
db_copper()
```

```python
db_silver()
```

```python
db_gold()
```
