```python
import yaml
import json
import pymysql

# Read YAML file
with open(r"C:\Users\Admin\Desktop\PYTHON\employee.yaml", "r") as fp:
    data = yaml.safe_load(fp)

# Read database credentials from config.json
with open(r"C:\Users\Admin\Desktop\PYTHON\config.json", "r") as fp:
    config = json.load(fp)

# Connect to MySQL database
conn = pymysql.connect(
    host=config["host"],
    port=config["port"],
    user=config["user"],
    password=config["password"],
    database=config["database"]
)

cur = conn.cursor()

# Insert YAML data into student table
for line in data:
    query = "insert into class values('{}','{}','{}')".format(line['id'], line['name'], line['city'])
    cur.execute(query)

conn.commit()
conn.close()
```
