```python

def file_read(filename):
    with open(filename) as fp:
        data = fp.read()
        return data

def count(filename):
    numeric = 0
    specialchar = 0

    characters = len(filename)
    words = len(filename.split())
    lines = len(filename.splitlines())

    for ch in filename:
        if ch.isnumeric():
            numeric += 1
        elif not ch.isalnum() and ch != " ":
            specialchar += 1
    
    result = {
    "characters": characters,
    "words": words,
    "numeric": numbers,
    "special_characters": specialchar,
    "lines": lines
    }

    return result  

def main():
    filename = "Sample.txt"
    fdata = file_read(filename)
    result = count(fdata)
    print(result)

if __name__ == '__main__':
    main()
```

 # Assignment 2

```python
def file_read(filename):
    fp = open(filename)
    data = fp.read()
    fp.close()
    return data

def mobile_filteration(filedata):
    valid_mobile = []
    invalid_mobile = []
    for words in filedata.split():
        if words.isnumeric():
            if len(words) == 10:
                if words.startswith('7') or words.startswith('8') or words.startswith('9'):
                    valid_mobile.append(words)
                else:
                    invalid_mobile.append(words)
    return valid_mobile, invalid_mobile

def file_write(filename, data):
    fp = open(filename, "w")
    for item in data:
        fp.write(item + '\n')
    fp.close()
    
def email_filterartion(filedata):
    valid_email = []
    invalid_email = []
    for words in filedata.split():
        if "@" in words:
            if words.endswith("@gmail.com"):
                valid_email.append(words)
            else:
                invalid_email.append(words)
    return valid_email, invalid_email

def main():
    filename = "sample.txt"
    fdata = file_read(filename)
    valid_mobile, invalid_mobile = mobile_filteration(fdata)
    valid_email, invalid_email = email_filterartion(fdata)
    file_write("valid_mobile.txt", valid_mobile)
    file_write("valid_email.txt", valid_email)
    file_write("invalid.txt", invalid_mobile + invalid_email)

if __name__ == "__main__":
    main()
```

 # Assignment 3

```python
def file_read(filename):
    fp = open(filename)
    data = fp.read()
    fp.close()
    return data


def char_specialisation(filedata):
    char = len(filedata)
    words = len(filedata.split())
    special_characters = []
    for ch in filedata:
        if not ch.isalnum() and not ch.isspace():
            special_characters.append(ch)
    return char, words, special_characters

def count_numeric(filedata):
    numeric = []
    for ch in filedata:
        if ch.isnumeric():
            numeric.append(ch)
    return numeric

def file_write(filename, data):
    fp = open(filename, "w")
    if isinstance(data, list):
        for item in data:
            fp.write(str(item) + "\n")
    else:
        fp.write(str(data) + "\n")
    fp.close()

def main():
    filename = "sample.txt"
    fdata = file_read(filename)
    char, words, special_characters = char_specialisation(fdata)
    numeric = count_numeric(fdata)
    file_write("character.txt", char)
    file_write("special_character.txt", special_characters)
    file_write("numeric.txt", numeric)
    print("Files created successfully.")


if __name__ == "__main__":
    main()
```
