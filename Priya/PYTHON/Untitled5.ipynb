```python
import json
import csv

def read_json_file(filename):
    with open("{}.json".format(filename), "r", newline="") as fp:
        return json.load(fp)
        
def get_columns(file):
    columns = {}
    for record in file:
        for key in record:
            columns[key] = 1
    return list(columns.keys())

def to_csv(keys, file):
    filename = ("{}.csv".format("emp2_output"))
    with open(filename, "w", newline="") as fp:
        writer = csv.DictWriter(fp, fieldnames=keys)
        writer.writeheader()
        writer.writerows(file)
        
if __name__ == '__main__':
    file = read_json_file(r"C:\Users\Admin\Desktop\PYTHON\emp2")
    keys = get_columns(file)
    to_csv(keys, file)
    print(keys)
```
