```python
emails=['test@gmail.com','sample@gmail.com','python@gmail.com','ml@gmail.com']
```

```python
def get_username(emp):
    return emp.split('@')[0]
```

```python
list(map(get_username,emails))
```

```python
list(map(lambda x:x.split('@')[0],emails))
```

```python
p = [1,2,45,756,45]
```

```python
def even_num(num):
    if num%2==0:
        return True
    else:
        return False
```

```python
list(filter(even_num,p))
```

```python
list(filter(lambda x:x%2==0,p))
```
