```python
import json
import csv

def read_json_file(filename):
    with open("{}.json".format(filename), "r", newline="") as fp:
        return json.load(fp)
        
def get_columns(file):
    columns = {}
    for record in file:
        for key in record:
            columns[key] = 1
    return list(columns.keys())

def to_csv(keys, file):
    filename = ("{}.csv".format("emp2_output"))
    with open(filename, "w", newline="") as fp:
        writer = csv.DictWriter(fp, fieldnames=keys)
        writer.writeheader()
        writer.writerows(file)
        
if __name__ == '__main__':
    file = read_json_file(r"C:\Users\Admin\Desktop\PYTHON\emp")
    keys = get_columns(file)
    to_csv(keys, file)
    print(keys)
```

# Yaml to CSV

```python
import csv
import yaml

def file_read(filename):
    with open("{}.yaml".format(filename ,"r")) as fp:
        return yaml.safe_load(fp)


def get_columns(file):
    columns = {}
    for records in file:
        for key in records:
            columns[key]=1
    return list(columns.keys())
        


def yaml_to_csv(keys, file):
    filename = "{}.csv".format("class_output1")
    with open(filename , "w", newline="") as fp:
        w=csv.DictWriter(fp, fieldnames=keys)
        w.writeheader()
        w.writerows(file)
    


def main():
    file = file_read(r"C:\Users\Admin\Desktop\PYTHON\employee")
    keys = get_columns(file)
    yaml_to_csv(keys,file)
    print(keys)

if __name__=='__main__':
    main()
    
    
```
