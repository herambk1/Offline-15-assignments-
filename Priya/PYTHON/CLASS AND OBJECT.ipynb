# Access Specifiers

```python
class Test:
    course = 'PDE'
    def m1(self):
        print("in Test class:",Test.course)
class Test1(Test):
    def m2(self):
        print("in child class:",Test.course)
class Test2(Test1):
    def m3(self):
        print("in grand child class:",Test.course)

obj = Test2()

obj.m1()
obj.m2()
obj.m3()
```

```python
class Test:
    course = 'PDE'
    _username = 'admin'
    def m1(self):
        print("in Test class:",Test.course)
        print("in Test class:",Test._username)
class Test1(Test):
    def m2(self):
        print("in child class:",Test.course)
        print("in Test class:",Test._username)
class Test2(Test1):
    def m3(self):
        print("in grand child class:",Test.course)
        print("in Test class:",Test._username)

obj = Test2()

obj.m1()
obj.m2()
obj.m3()
```

```python
class Test:
    course = 'PDE'
    _username = 'admin'
    __password = 'admi@123'
    def m1(self):
        print("in Test class:",Test.course)
        print("in Test class:",Test._username)
        print("in same class:",Test.__password)
class Test1(Test):
    def m2(self):
        print("in child class:",Test.course)
        print("in Test class:",Test._username)
        print("in same class:",Test.__password)
class Test2(Test1):
    def m3(self):
        print("in grand child class:",Test.course)
        print("in Test class:",Test._username)
        print("in same class:",Test.__password)

obj = Test2()

obj.m1()
obj.m2()
obj.m3()
```
