```python
import yaml
import csv


def read_yaml_file(filename):

    with open("{}.yaml".format(filename), "r") as fp:
        return yaml.safe_load(fp)


def get_columns(file):

    columns = {}

    for record in file:

        for key in record:
            columns[key] = 1

    return list(columns.keys())


def to_csv(keys, file):

    filename = "{}.csv".format("emp1_output")

    with open(filename, "w", newline="") as fp:

        writer = csv.DictWriter(
            fp,
            fieldnames=keys
        )

        writer.writeheader()

        for record in file:

            if isinstance(record, dict):
                writer.writerow(record)

            else:
                print("Invalid record:", record)


if __name__ == '__main__':

    file = read_yaml_file(
        r"C:\Users\Admin\Desktop\PYTHON\employee"
    )

    print("Data:", file)
    print("Type:", type(file))

    keys = get_columns(file)

    to_csv(keys, file)

    print(keys)
```
