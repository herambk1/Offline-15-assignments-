```python
def read_file(filename):
    with open(filename,"r")as file:
        return file.read()
  

def char_file(data):
    with open(r"C\User\Admin\Desktop\char_file.txt", "w") as file:
        for ch in data:
            if ch.isalpha():
                file.write(ch)

def num_file(data):
    with open(r"C\User\Admin\Desktop\num_file.txt", "w") as file:
        for ch in data:
            if ch.isnumeric():
                file.write(ch)

def special_char_file(data):
    with open(r"C\User\Admin\Desktop\Special_char.txt","w") as file:
        for ch in data:
            if not ch.isalnum() and not ch.isspace():
                file.write(ch)


if __name__ == "__main__":
    
    data = read_file(r"C\User\Admin\Desktop\source.txt")
    char_file(data)
    num_file(data)
    special_char_file(data)
```
