```python
import json
import csv

# JSON string
json_string = '''
[
    {"id": 101, "name": "Priya", "dept": "IT", "salary": 50000},
    {"id": 102, "name": "Rahul", "dept": "HR", "salary": 60000},
    {"id": 103, "name": "Amit", "dept": "IT", "salary": 70000}
]
'''

# Convert JSON string into Python list
data = json.loads(json_string)

# Create CSV file
with open("employees.csv", "w", newline="") as fp:

    fieldnames = data[0].keys()

    writer = csv.DictWriter(
        fp,
        fieldnames=fieldnames)
    writer.writeheader()

    # Write records
    writer.writerows(data)

print("JSON converted into CSV successfully")
```

```python
employees = [
    ("Amit", "IT", 45000),
    ("Raj", "HR", 55000),
    ("Priya", "IT", 75000),
    ("Neha", "Finance", 65000),
    ("Suresh", "IT", 90000),
    ("Anita", "HR", 40000)
]
filtered = filter(lambda x: x[1] == "IT" and x[2]>50000,employees)
updated = map(lambda x:(x[0], x[1],x[2]*1.10),filtered)
result = sorted(updated,key =lambda x: x[2],reverse =True)
print(result)
```

```python
students = [
    ("Amit", 85),
    ("Raj", 45),
    ("Priya", 92),
    ("Neha", 67),
    ("Suresh", 38),
    ("Anita", 78)
]

filtered = filter(
    lambda x: x[1] >= 60,
    students
)

updated = map(
    lambda x: (x[0], x[1] + 5),
    filtered
)

sorted_students = sorted(
    updated,
    key=lambda x: x[1],
    reverse=True
)

result = list(
    map(
        lambda x: f"{x[0]} - {x[1]}",
        sorted_students
    )
)

print(result)
```
