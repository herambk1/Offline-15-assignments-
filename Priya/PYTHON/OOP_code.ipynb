```python
import paramiko

# Define server credentials
HOST = 'eu-central-1.sftpcloud.io'
USERNAME = 'bfac07f61ef14d989d961db2736a7df4'
PASSWORD = 'j2ftug9rQIhT4joh2XzoCKwxTi8lhHss'# Or use private_key='/path/to/key'
PORT = 22

# Create an SSH client instance
ssh = paramiko.SSHClient()

# Automatically add unknown host keys (Testing / Dev environment ONLY)
ssh.set_missing_host_key_policy(paramiko.AutoAddPolicy())

try:
    # Connect to the server
    ssh.connect(HOST, port=PORT, username=USERNAME, password=PASSWORD)
    
    # Open an SFTP session
    sftp = ssh.open_sftp()
    print("Paramiko SFTP connection established!")
    
    # Perform file operations

    files =  sftp.listdir(".")

    for file in files:
        if file.endswith('.pdf'):
            sftp.get("{}".format(file),'sftpfiles/{}'.format(file))
    # Close SFTP session
    sftp.close()

finally:
    # Always close the core SSH connection
    ssh.close()
```

```python
import paramiko

HOST = "eu-central-1.sftpcloud.io"
USERNAME = "your_username"
PASSWORD = "your_new_password"
PORT = 22


def connect_sftp():
    ssh = paramiko.SSHClient()

    # Testing/Development only
    ssh.set_missing_host_key_policy(paramiko.AutoAddPolicy())

    try:
        ssh.connect(
            hostname=HOST,
            port=PORT,
            username=USERNAME,
            password=PASSWORD
        )

        print("SFTP connection successfully established!")

        sftp = ssh.open_sftp()

        # List files from SFTP server
        files = sftp.listdir()

        print("Files on SFTP server:")

        for file in files:
            print(file)

        sftp.close()
        ssh.close()

    except Exception as e:
        print("SFTP connection failed:", e)


if __name__ == "__main__":
    connect_sftp()
```
