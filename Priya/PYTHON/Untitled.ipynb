```python
def fabbonacci_series():
    a,b=0,1
    while a<=100:
        yield a
        a,b=b,a+b
```

```python
obj=fabbonacci_series()
```

```python
for num in obj:
    print(num)
```

```python
#Armstrog
```

```python
number=input("Enter the number:")
len_no=len(number)
total=0
for num in number:
    total=total + int(num)**len_no
if total == int(number):
        print("Armstrong")
else:
        print("Not armstrong")
```
