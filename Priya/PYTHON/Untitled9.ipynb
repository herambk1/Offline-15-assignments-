```python
def prime_number(num):
    if num <=1:
        print('not prime')
    else:
        for i in range(2,num):
            if num%i==0:
                return "not prime"
                break
        else:
            return "prime number"
        
        
    
```

```python
a=prime_number(6)
print(a)
```

```python
def palindrome(num):
   
    reverse=""

    for i in range(len(num)-1,-1,-1):
        reverse=reverse+num[i]

    if num==reverse:
        print("number is palindrome")
    else:
        print("number is not palindrome")

num=input("enter the number:")
palindrome(num)
```
