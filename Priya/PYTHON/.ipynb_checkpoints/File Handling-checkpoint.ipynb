```python
def file_read(filename):
    fp=open(filename)
    data=fp.read()
    return data

def filter_mobile_num(filedata):
    valid_mobilenos=[]
    invalid_mobilenos=[]
    for word in filedata.split():
        if word.isnumeric():
            if len(word)==10:
                if word.startswith('7') or word.startswith('9'):
                    valid_mobilenos.append(word)
                else:
                    invalid_mobilenos.append(word)
    return valid_mobilenos,invalid_mobilenos

def filter_email_ids(filedata):
    valid_email=[]
    invalid_email=[]
    for word in filedata.split():
        if  '.com' in word:
            if word.endswith("@gmail.com"):
                valid_email.append(word)
            else:
                invalid_email.append(word)
    return valid_email,invalid_email

    
def main():
    filename='sample.txt'
    fdata=file_read(filename)
    v_mobile,in_mobile=filter_mobile_num(fdata)
    v_email,in_email=filter_email_ids(fdata)
    print(v_mobile)
    print(in_mobile)
    print(v_email)
    print(in_email)
if __name__=='__main__':
    main()
```

```python
# Read sample.txt
def 
with open("sample.txt", "r") as file:
    data = file.read()


# 1. Count Characters
char = len(data)


# 2. Count Words
words = len(data.split())


# 3. Count Special Characters
special_characters = 0

for ch in data:
    if not ch.isalnum() and not ch.isspace():
        special_characters += 1


# 4. Count Lines
lines = len(data.splitlines())


# 5. Count Numbers
numbers = 0

for ch in data:
    if ch.isdigit():
        numbers += 1

# Store all results in dictionary
result = {
    "Char": char,
    "Words": words,
    "SpecialCharacters": special_characters,
    "Lines": lines,
    "Numbers": numbers
}

# Create output file
with open("sample_output.txt", "w") as file:
   
    file.write("Char : " + str(char) + "\n")
    file.write("Words : " + str(words) + "\n")
    file.write("SpecialCharacters : " + str(special_characters) + "\n")
    file.write("Lines : " + str(lines) + "\n")
    file.write("Numbers : " + str(numbers) + "\n")

print("Output file created successfully.")
```
