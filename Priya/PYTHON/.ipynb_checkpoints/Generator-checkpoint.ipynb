```python
def fabbonacci_series():
    a,b=0,1
    while a<=100:
        yield a
        a,b=b,a+b
```

```python
obj=fabbonacci_series()
```

```python
for num in obj:
    print(num)
```

```python
def mygen(start,end):
    i =start
    while i<=end:
        yield i
        i+=1
```

```python
obj=mygen(5,500)
```

```python
for i in obj:
    print(i)
```

```python
p=['56','7','7','76']
```

```python
def convert_into_int_rev(plist):
    pl=[]
    for element in plist:
        pl.append(int(element))
    pl.reverse()
    return pl
```

```python
convert_into_int_rev(p)
```
