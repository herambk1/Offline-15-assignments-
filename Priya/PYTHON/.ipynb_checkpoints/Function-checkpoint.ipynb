```python
def take_inputs_from_user():
    s1=int(input("Enter the Marks:"))
    s2=int(input("Enter the Marks:"))
    s3=int(input("Enter the Marks:"))
    emailid=input("Enter the email id:")
    return s1,s2,s3,emailid

def calculate_total_avg(a,b,c):
    total= a+b+c
    avg= total/3
    return total,avg
    
def calculate_grade(av):
    if av>75:
        return "Outstanding"
    else:
        return "Normal"
        
def is_email_valid(em):
    if em.endswith("@gmail.com"):
        return True
    else:
        return False

def main():
    sub1,sub2,sub3,emailid=take_inputs_from_user()
    total,avg=calculate_total_avg(sub1,sub2,sub3)
    grade=calculate_grade(avg)
    status=is_email_valid(emailid)
    if status:
        print("email id valid")
        print("sub marks:",sub1,sub2,sub3)
        print("Total:",total)
        print("Avg:",avg)
        print("Grade",grade)
    else:
        print("email id invalid")

if __name__ == '__main__':
    main()
   
```
