```python
class Emp:
    companyname = 'IBM'
    def __init__(self,id,name,city):
        self.id = id
        self.name = name
        self.city = city
    def show(self):
        print("Company Name:",Emp.companyname)
        print("ID:",self.id)
        print("Name:",self.name)
        print("City:",self.city)

    @classmethod
    def change_companyname(cls):
        cls.companyname = 'Infosys'


Emp.change_companyname() 
obj = Emp(101,'Arya','Pune')
obj.show()

obj1 = Emp(102,'Umesh','Mumbai')
obj1.show()
```

```python
class Test:
    @staticmethod
    def add(x,y):
        z = x + y
        return z
obj = Test()
obj.add(100,200)
```

```python
class A:
    def m1(self):
        print("in m1")
class B(A):
    def m2(self):
        print("in m2")
obj =B()
obj.m1()
obj.m2()
```

```python
class A:
    def m1(self):
        print("From A")
class B(A):
    def m1(self):
        print("From B")
class C(A):
    def m1(self):
        print("From C")
class D(B,C):
    def m1(self):
        print("From D")
obj =D()
obj.m1()
```

```python
class A:
    def m1(self):
        print("from A")
class B(A):
    def m1(self):
        super().m1()
        print("from B")
```

```python
obj =B()
```

```python
obj.m1()
```

```python
class PersonalDetails:
    def __init__(self,id,name,city):
        self.id= id
        self.name = name
        self.city = city

    def show(self):
        print("ID:",self.id)
        print("Name:",self.name)
        print("City:",self.city)

class PanAdded(PersonalDetails):
    def __init__(self,id,name,city,pan):
        super().__init__(id,name,city)
        self.pan = pan

    def show(self):
        super().show()
        print("PAN:",self.pan)
        
class PassportAdded(PanAdded):
    def __init__(self,id,name,city,pan,passport):
        super().__init__(id,name,city,pan)
        self.passport = passport

    def show(self):
        super().show()
        print("Passport:",self.passport)
        
        
    
        
```

```python
obj=PassportAdded(101,'Arya','Pune','ddyghgd',134356)
```

```python
obj.show()
```

```python
class A:
    def m1(self):
        print("from A")
class B(A):
    def m1(self):
        super().m1()
        print("from B")
class C(A):
    def m1(self):
        print("from C")
class D(B,C):
    def m1(self):
        super().m1()
        print("from D")
```

```python
obj = D()
```

```python
obj.m1()
```
