```python
import pymysql
import json

# Read JSON file
with open(r"C:\Users\Admin\Desktop\PYTHON\student.json", "r") as fp:
    data = json.load(fp)

# Read database credentials
with open(r"C:\Users\Admin\Desktop\PYTHON\config.json", "r") as fp:
    config = json.load(fp)

# Ask user for database and table
database_name = input("Enter database name: ")
table_name = input("Enter table name: ")

# Connect to MySQL
conn = pymysql.connect(
    host=config["host"],
    port=config["port"],
    user=config["user"],
    password=config["password"],
    database=database_name
)

cur = conn.cursor()

# Insert data
for line in data:
    query = "INSERT INTO {} VALUES ({}, '{}', '{}')".format(
        table_name,
        line['id'],
        line['name'],
        line['city']
    )

    print(query)
    cur.execute(query)

conn.commit()
conn.close()

print("Data inserted successfully")
```
