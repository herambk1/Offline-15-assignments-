```python
lst = [9, 8, 10, 7]
```

```python
larg = lst[0]
s, t = 0, 0
for i in lst:
    if i > larg:
        s= larg
        larg = i
    elif i > s:
        t = s
        s = i
    elif i > t:
        t = i
```

```python
prime = int(input("Enter num. "))
cnt = 0
for i in range(2,prime+1):
    if prime % i == 0:
        cnt += 1

if cnt == 1: 
    print("Number is prime")
else :
    print("Number is not prime")
```

```python
def email_validation(email):
    if "@" not in email:
        return "False 1"

    parts = email.split("@")

    if len(parts) != 2:
        return "False 2"

    username, domain = parts

    if not username or not domain:
        return "False 3"

    if "." not in domain:
        return "False 4"

    if domain.startswith(".") or domain.endswith("."):
        return "False 5"

    return "is Valid"

def main():
    email = "sakshar@gm.ailcom"
    result = email_validation(email)
    print(result)

if __name__ == '__main__':
    main()
```
