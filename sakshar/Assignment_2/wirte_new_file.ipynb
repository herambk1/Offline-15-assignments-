```python
# write characters, special char, numeric in new file 

def no_of_char(data):
    with open("characters_file.txt","w") as fp:
        for char in data:
            fp.write(char)

def no_of_special_char(data):
    with open("special_char_file.txt","w") as fp:
        for char in data:
            if not str(char).isalnum():
                fp.write(char)

def no_of_numric(data):
    with open("noumber_file.txt","w") as fp:
        for char in data:
            if str(char).isnumeric():
                fp.write(char)

def main():
    with open("data.txt") as fp:
        data = fp.read()

        no_of_char(data)
        no_of_special_char(data)
        no_of_numric(data)
        

if __name__ == '__main__':
    main()
```
