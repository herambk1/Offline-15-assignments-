```python
def read_file(filename):
    fp=open(filename,'r')
    data=fp.read()
    return data

def file_write(filename,data):
    fp=open(filename,'w')
    for item in data:
        fp.write(item +'\n')
    fp.close()
     
def validate_mob_num(data):
    valid_mob=[]
    invalid_mob=[]
    for word in data.split():
        if word.isnumeric():
            if len(word)==10 and word[0] in '789':
                    valid_mob.append(word)
            else:
                    invalid_mob.append(word)
    return valid_mob,invalid_mob


def validate_email(data):
    valid_email=[]
    invalid_email=[]
    for word in data.split():
        if word.count('@')==1:
            if word.endswith('@gmail.com'):
                valid_email.append(word)
            else:
                invalid_email.append(word)
    return valid_email,invalid_email

def invalid_data(data):
    invalid_file=[]
    for word in data.split():
        if not word.isnumeric() and not word.endswith('@gmail.com'):
            invalid_file.append(word)
    return invalid_file
            

def main():
    file='sample.txt'
    
    fdata=read_file(file)
    
    valid_mob,invalid_mob=validate_mob_num(fdata)
    valid_email,invalid_email=validate_email(fdata)
    invalid_Data=invalid_data(fdata)
    
    file_write("valid_mobile.txt",valid_mob)
    file_write("invalid_mobile.txt",invalid_mob)
    file_write("valid_Email.txt",valid_email)
    file_write("invalid_Email.txt",invalid_email)
    file_write("invalid_data.txt",invalid_Data)

if __name__ == '__main__':
    main()
```

## Assignment-2

```python
def read_file(filename):
    fp = open(filename, 'r', encoding='utf-8', errors='ignore')
    data = fp.read()
    fp.close()
    return data


def char_count(data):
    char = {}

    for ch in data:
        if ch.isalpha():
            if ch in char:
                char[ch] += 1
            else:
                char[ch] = 1

    return char


def words_count(data):
    word_count = {}

    for word in data.split():
        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1

    return word_count


def spcl_count(data):
    spcl = {}

    for ch in data:
        if not ch.isalnum() and not ch.isspace():
            if ch in spcl:
                spcl[ch] += 1
            else:
                spcl[ch] = 1

    return spcl


def line_count(data):
    lines_count = {}

    for line in data.splitlines():
        line = line.strip()

        if line:
            if line in lines_count:
                lines_count[line] += 1
            else:
                lines_count[line] = 1

    return lines_count


def numbers_count(data):
    number_count = {}

    for ch in data:
        if ch.isnumeric():
            if ch in number_count:
                number_count[ch] += 1
            else:
                number_count[ch] = 1

    return number_count


def main():
    file = 'sample123.txt'

    fdata = read_file(file)

    word_count = words_count(fdata)
    lines_count = line_count(fdata)
    character_count = char_count(fdata)
    special_count = spcl_count(fdata)
    number_count = numbers_count(fdata)

    print('Word count:', word_count)
    print('Line count:', lines_count)
    print('Character count:', character_count)
    print('Special count:', special_count)
    print('Number count:', number_count)


if __name__ == '__main__':
    main()
```
