```python
import csv
import os

def fetch_file(source):
    dir=os.listdir(source)
    csv_files=[]
    for file in dir:
        if file.endswith(".csv"):
            csv_files.append(file)
    return csv_files

def read_csv(source,csv_file):
    with open("{}/{}".format(source,csv_file),'r') as fp:
        reader =fp.read()
    return reader


def write_into_destination(destination,csv_file,fdata):
    with open("{}/{}".format(destination,csv_file),'w') as fp:
        writer = fp.write(fdata)

def delete_from_source(source,csv_file):
    os.remove("{}/{}".format(source,csv_file))

def main():
    source=r"C:\Users\DABC\Desktop\Python AIN"
    destination=r"C:\Users\DABC\Desktop\Python AIN\Destination"
    csv_files=fetch_file(source)
    print(csv_files)
    for csv_file in csv_files:
        fdata=read_csv(source,csv_file)
        print(fdata)
        write_into_destination(destination,csv_file,fdata)
        delete_from_source(source,csv_file)

if __name__=="__main__":
    main()
```
