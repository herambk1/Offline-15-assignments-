````python
```python
def separate_data():
    with open("input.txt", "r") as file:
        data = file.readlines()

    with open("emails.txt", "a") as email_file, \
         open("mobile.txt", "a") as mobile_file, \
         open("alphanumeric.txt", "a") as alpha_file:

        for line in data:
            line = line.strip()

            if "@" in line and "." in line:
                email_file.write(line + "\n")

            elif line.isdigit() and len(line) == 10:
                mobile_file.write(line + "\n")

            elif line.isalnum() and not line.isdigit():
                alpha_file.write(line + "\n")


separate_data()
```
````
