```python
# check the no is prime or not
no = int(input("Enter the number: "))

if no >= 2:
    prime = True

    for i in range(2, no):
        if no % i == 0:
            prime = False
            break

    if prime:
        print("Prime")
    else:
        print("Not Prime")

else:
    print("Valid number enter")
```

```python
# remove duplicate ele form list
def remove_duplicate(numbers):
    unique=[]
    for i in  numbers:
        if i not in unique:
            unique.append(i)
    return unique

numbers=[1,5,1,2,2,4,7,8,9,7,7,7,522,56,100,100,233,2,5]

result=remove_duplicate(numbers)

print(result)
```

```python
# check the frequency of number
def check_frequency(numbers):
    freq={}
    for key in numbers:
         if key in freq:
             freq[key]+=1
         else:
             freq[key]=1
    return freq 
     
         
numbers=[1,5,1,2,2,4,7,8,9,7,7,7,522,56,100,100,233,2,5]

result=check_frequency(numbers)
print(result)
```

```python
# find common ele form two list 
def common_element(p,q):
    common=[]
    for num in p:
        if num in q:
            if num not in common:
                common.append(num)
    return common


p=[1,2,3,4,5,6,7,8]
q=[1,2,3,4,5]

result=common_element(p,q)
print(result)
```

```python
# find common ele form two list comprehension

def common_element(p,q):
    return [num for num in p if num in  q]

p=[1,2,3,4,5,6,7,8]
q=[1,2,3,4,5]

result=common_element(p,q)
print(result)
```

```python
# dictionray from two list
def new_dict(p,q):
    data={}
    for i in range(len(p)):
        data[p[i]]=q[i]
    return data



p = ["name", "age", "Role", "Package"]
q = ["Yograni", 28, "DE", 2500000]

result=new_dict(p,q)
print(result)
```

```python
# dictionray from two list
def new_dict(p,q):
    return dict(zip(p,q))

p = ["name", "age", "Role", "Package"]
q = ["nisha", 28, "DE", 2500000]

result=new_dict(p,q)
print(result)
```

```python
def check_ipaddress(ipaddress):
    parts=ipaddress.split('.')

    if len(parts)!=4:
        return "Invaldi IP"
        
    for digit in parts:
        if not digit.isnumeric():
            return "Not a number"
        if int(digit) < 0 or int(digit) > 255:
            return "Invalid - not in range"
        return "Valid IP"

ipaddress=input("enter the IP address")

result=check_ipaddress(ipaddress)
print(result)
```
