```python
import json 
import csv

def readJsonFile(file):
    with open(file,"r") as fp:
        data=json.load(fp)
    return data


def create_csv(data):
    head=[]
    for key in data[0]:
        head.append(key)
        
    with open("jsontocsvFile.csv", "w", newline="") as fp:
        writer = csv.DictWriter(fp, fieldnames=head)
        writer.writeheader()
        writer.writerows(data)
        

def main():
    jsonFile=readJsonFile("jdata.json")
    create_csv(jsonFile)

if __name__ == "__main__":
    main()
```
