```python
def extractFileInfo(fileName):
    numbers=0
    charchters=0
    special_chars=0
    lines=0
    words=0

    with open(fileName,"r") as file:
        # no of line
        for line in file:
            lines+=1

            # count words
            words+=len(line.split())
        # count no ch specchar
            for ch in line:
                if ch.isnumeric():
                    numbers+=1
                elif ch.isalpha():
                    charchters+=1
                elif not ch.isspace():
                    special_chars+=1

            
    return{
        "lines":lines,
        "words":words,
        "numbers":numbers,
        "charchters":charchters,
        "specialChar":special_chars
          
          }

def createFile(data):
    with open("test_output.txt","w") as fp:
        fp.write(str(data))
    
    
def main():
    data=extractFileInfo("test.txt")
    result=createFile(data)
    print(data)

if __name__ == "__main__":
    main()
    
```
