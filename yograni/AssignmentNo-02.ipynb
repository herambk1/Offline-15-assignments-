```python
def readFile(file):
    with open(file,"r") as fp:
        data= fp.read()

    return data
   

def mobileNo(data):
    vaildMobileNo=[]
    invaildMobileNo=[]

    for num in data.split():
        if num.isnumeric():
            if len(num)==10:
                if num.startswith(("9","8","7")):
                    vaildMobileNo.append(int(num))
                else:
                    invaildMobileNo.append(int(num))
    return vaildMobileNo,invaildMobileNo
                
  
def emailID(data):
    vaildEmailId=[]
    invaildEmailID=[]
    for words in data.split():
        if "@" in words:
            if words.endswith(("@gmail.com" , "@yahoo.com")):
                vaildEmailId.append(words)
            else:
                invaildEmailID.append(words)
    return vaildEmailId,invaildEmailID

def createFile(fileName,data):
    with open(fileName,"w")as fp:
        for item in data:
            fp.write(str(item)+ "\n")



def main(): 
    fileName="new.txt"
    data=readFile(fileName)
    validMobileNo,invaildMobileNo= mobileNo(data)
    vaildEmailId,invaildEmailID=emailID(data)

    createFile("vaild_mobile.txt",validMobileNo)
    createFile("inVaild_mobile.txt",invaildMobileNo)
    createFile("vaild_email.txt",vaildEmailId)
    createFile("inVaild_email.txt",invaildEmailID)
    
    print(validMobileNo, invaildMobileNo)
    print(vaildEmailId ,invaildEmailID)

if __name__  == "__main__":
    main()

```
