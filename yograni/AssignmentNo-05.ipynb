```python
import yaml
import csv

def read_Ymal_File(fileName):
    with open(fileName,"r") as fp:
        data=yaml.safe_load(fp)
    return data

def write_csv_file(data):
    
    with open("yaml_to_csv.csv","w",newline="") as fp:
        w=csv.DictWriter(fp,fieldnames=data[0].keys())
        w.writeheader()
        w.writerows(data)
    print("CSV file created successfully")

def main():
    fileName="data.yaml"
    data=read_Ymal_File(fileName)
    write_csv_file(data)

if __name__ =="__main__":
    main()
```
