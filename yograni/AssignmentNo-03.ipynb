```python
Assignement No.03
If we have 1 csv file with various departments 
from that create separate file for each department and which is not hardcoded should be dynamic
```

```python
import csv

def readFile(file):
    with open(file,"r")as fp:
        reader=csv.DictReader(fp)
        data=list(reader)
    return data

def group_by_department(data):
    dept_data={}
    for line in data:
        dept=line["department"]

        if dept not in dept_data:
            dept_data[dept]=[]

        dept_data[dept].append(line)
    return dept_data

def write_files(filterdept):
    
    for dept, data in filterdept.items():
        fileName=f"{dept}.csv"

        with open(fileName,"w",newline="") as fp:
            fieldName=data[0].keys()
            w=csv.DictWriter(fp,fieldnames=fieldName)
            w.writeheader()
            w.writerows(data)

def main():
    fileName=r"D:\Downloads\sample-employees.csv"
    data=readFile(fileName)
    filterdept=group_by_department(data)
    write_files(filterdept)
    print("success")
   
if __name__ == "__main__":
    main()


 # departments=set()
    # for line in data:
    #     dept=line["department"]
    #     departments.add(dept)
    # return departments

# def filter_data(depts, data):
#     department_data = {}

#     for dept in depts:
#         department_data[dept] = []

#         for line in data:
#             if line["department"] == dept:
#                 department_data[dept].append(line)

#     return department_data
```
